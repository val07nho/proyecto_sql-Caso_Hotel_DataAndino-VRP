```python

%sql
DESCRIBE TABLE cursodataacademy.proyecto_sql_vnrp.hotel_bookings;
```

```python
%sql
SELECT
    COUNT(*) AS total_reservas,

    SUM(CASE WHEN is_canceled = 1 THEN 1 ELSE 0 END)
        AS reservas_canceladas,

    SUM(CASE WHEN is_canceled = 0 THEN 1 ELSE 0 END)
        AS reservas_no_canceladas
FROM cursodataacademy.proyecto_sql_vnrp.hotel_bookings;
```

```python
%sql
SELECT
    SUM(CASE WHEN hotel IS NULL THEN 1 ELSE 0 END)
        AS nulos_hotel,

    SUM(CASE WHEN distribution_channel IS NULL THEN 1 ELSE 0 END)
        AS nulos_canal,

    SUM(
        CASE
            WHEN country IS NULL
              OR TRIM(country) = ''
              OR LOWER(TRIM(country)) IN ('nan', 'null')
            THEN 1 ELSE 0
        END
    ) AS pais_sin_dato,

    SUM(
        CASE
            WHEN children IS NULL
              OR LOWER(CAST(children AS STRING)) = 'nan'
            THEN 1 ELSE 0
        END
    ) AS hijos_sin_dato,

    SUM(CASE WHEN babies IS NULL THEN 1 ELSE 0 END)
        AS nulos_bebes,

    SUM(CASE WHEN adr IS NULL THEN 1 ELSE 0 END)
        AS nulos_adr,

    SUM(CASE WHEN lead_time IS NULL THEN 1 ELSE 0 END)
        AS nulos_anticipacion,

    SUM(CASE WHEN total_of_special_requests IS NULL THEN 1 ELSE 0 END)
        AS nulos_solicitudes,

    SUM(CASE WHEN booking_changes IS NULL THEN 1 ELSE 0 END)
        AS nulos_cambios,

    SUM(
        CASE
            WHEN arrival_date_year IS NULL
              OR arrival_date_month IS NULL
              OR arrival_date_day_of_month IS NULL
            THEN 1 ELSE 0
        END
    ) AS llegada_incompleta,

    SUM(CASE WHEN reservation_status_date IS NULL THEN 1 ELSE 0 END)
        AS nulos_fecha_estado

FROM cursodataacademy.proyecto_sql_vnrp.hotel_bookings;
```

```python
%sql
CREATE OR REPLACE TABLE cursodataacademy.proyecto_sql_vnrp.tb_reservas AS 
SELECT *
FROM cursodataacademy.proyecto_sql_vnrp.hotel_bookings;
```
