# Hipótesis 1: 
¿Los hijos aseguran el viaje? Escribe una consulta usando CASE WHEN que clasifique a la reserva como "Con Hijos" (si children > 0 o babies > 0) o "Sin Hijos". Agrupa por esta nueva categoría y calcula la tasa de cancelación (puedes sacar el promedio - AVG - de la columna is_canceled). ¿Qué segmento cancela más?


```python
%sql
SELECT 
    CASE 
        WHEN babies>0 OR children>0 THEN 'Con Hijos'
        ELSE 'Sin Hijos'
    END AS segmento_cancelacion,
    COUNT(*) AS total_reservas,
    ROUND(AVG(is_canceled) * 100, 2) AS tasa_cancelacion_pct
FROM cursodataacademy.proyecto_sql_vnrp.tb_reservas
GROUP BY segmento_cancelacion
ORDER BY tasa_cancelacion_pct DESC;
```

**Respuesta Hipotesis 1**

Las reservas con hijos registran 2,30 puntos porcentuales menos de cancelación que las reservas sin hijos (34,92 % frente a 37,22 %). Aunque presentan menor riesgo observado, cerca de una de cada tres se cancela; por ello, este criterio por sí solo no justifica flexibilizar las políticas de garantía del hotel.

# Hipótesis 2:
 El impacto de las solicitudes especiales. Agrupa por la columna total_of_special_requests y calcula la tasa de cancelación (AVG(is_canceled)) para cada cantidad de solicitudes (0, 1, 2, etc.). Ordena por la cantidad de solicitudes de forma ascendente. ¿A medida que el cliente pide más cosas, cancela más o cancela menos?




```python
%sql
SELECT
    total_of_special_requests,
    COUNT(*) AS total_reservas,
    ROUND(AVG(is_canceled)*100,2) AS tasa_cancelacion_pct
FROM cursodataacademy.proyecto_sql_vnrp.tb_reservas
GROUP BY total_of_special_requests
ORDER BY total_of_special_requests ASC;
```

**Respuesta Hipotesis 2**


Las reservas sin solicitudes especiales presentan la mayor tasa de cancelación (47,72 %), frente a tasas de entre 5 % y 22,10 % cuando existe alguna solicitud. Se observa una asociación general entre más solicitudes y menor cancelación, aunque no es una reducción estrictamente progresiva se confirma la hipotesis.

# Hipótesis 3:
 ¿Cambiar la reserva es señal de compromiso? Al igual que en los ejercicios anteriores, utiliza CASE WHEN para segmentar la columna booking_changes en tres grupos: "Sin Cambios" (0), "1 a 3 cambios" y "Más de 3 cambios". Calcula la tasa de cancelación para cada grupo. ¿Quiénes tienen menor riesgo de cancelar?


```python
%sql
SELECT
    CASE
        WHEN booking_changes = 0 THEN 'Sin Cambios'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
        WHEN booking_changes > 3 THEN 'Más de 3 cambios'
        ELSE 'Sin información'
    END AS grupo_cambios,
    COUNT(*) AS total_reservas,
    ROUND(AVG(is_canceled) * 100, 2) AS tasa_cancelacion_pct
FROM cursodataacademy.proyecto_sql_vnrp.tb_reservas
GROUP BY grupo_cambios
ORDER BY tasa_cancelacion_pct ASC;
```

**Respuesta hipotesis 3**

Los resultados respaldan la hipótesis: las reservas con modificaciones presentan menor cancelación, especialmente aquellas con 1 a 3 cambios (15,59 % frente al 40,85 % sin cambios). Sin embargo, más cambios no implican necesariamente menor riesgo. Se recomienda utilizar este indicador junto con otros factores para priorizar el seguimiento de reservas.

# Hipótesis 4:
