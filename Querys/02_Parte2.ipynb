El Costo del Marketing Inútil (Agregación con Filtro): Sabiendo que a la agencia (canal TA/TO) se le paga USD 1.5 por cada reserva. ¿Cuánto dinero en total se ha pagado a la agencia por reservas que finalmente fueron canceladas (is_canceled = 1)? Escribe la consulta que calcule este monto de pérdida.


```python
%sql
SELECT
    COUNT(*) AS reservas_canceladas_agencia,
    COUNT(*) * 1.5 AS perdida_comisiones_usd
FROM cursodataacademy.proyecto_sql_vnrp.tb_reservas
WHERE distribution_channel = 'TA/TO' AND is_canceled = 1;
```

**Conclusión:** Se pagaron USD 60 228 en comisiones por 40 152 reservas del canal TA/TO que finalmente fueron canceladas. Este monto sustenta renegociar el contrato para vincular las comisiones a reservas efectivamente concretadas.


**El Costo de Cancelar Tarde (DATE_DIFF): (Considera que tienes las columnas arrival_date y reservation_status_date ambas en formato DATE). Cuando un huésped cancela con menos de 3 días de anticipación respecto a su llegada, el hotel pierde USD 120. Escribe una consulta que cuente cuántas reservas cumplen esta condición (is_canceled = 1 y diferencia entre llegada y cancelación menor a 3 días) y multiplica esa cantidad por 120 para ver la pérdida total.**

```python
%sql
SELECT
    COUNT(*) AS reservas_canceladas_tarde,
    COUNT(*) * 120 AS perdida_total
FROM cursodataacademy.proyecto_sql_vnrp.tb_reservas
WHERE is_canceled = 1
  AND DATEDIFF(
      arrival_date,
      TO_DATE(reservation_status_date, 'd/M/yy')
  ) < 3;
```

**Concluiones** Las 3 204 cancelaciones realizadas con menos de tres días de anticipación representan un costo operativo estimado de USD 384 480. Se recomienda evaluar garantías y mecanismos de confirmación previa para reducir la exposición a cancelaciones de última hora.

```python
%sql
SELECT
    COUNT(country) AS cant_registros,
    COUNT(DISTINCT country) AS cant_unica_registros,
    CASE
        WHEN COUNT(country) - COUNT(DISTINCT country) = 0 THEN 'EL CAMPO ES UNICO'
        ELSE 'TIENE VALORES REPETIDOS'
    END AS estatus_campo
FROM cursodataacademy.proyecto_sql_vnrp.tb_reservas;
```

```python
%sql
SELECT
    COUNT(cod_pais) AS cant_registros,
    COUNT(DISTINCT cod_pais) AS cant_unica_registros,
    CASE
        WHEN COUNT(cod_pais) - COUNT(DISTINCT cod_pais) = 0 THEN 'EL CAMPO ES UNICO'
        ELSE 'TIENE VALORES REPETIDOS'
    END AS estatus_campo
FROM cursodataacademy.proyecto_sql_vnrp.tb_paises;
```

```python
%sql
SELECT
    P.nombre_pais,
    R.is_canceled
FROM cursodataacademy.proyecto_sql_vnrp.tb_reservas R
LEFT JOIN cursodataacademy.proyecto_sql_vnrp.tb_paises P
ON R.country = p.cod_pais
LIMIT 10;
```

**Conclusiones** El cruce relaciona el estado de cancelación con el nombre del país, conservando las reservas aunque no tengan coincidencia en el catálogo. Esto permite preparar el análisis geográfico sin excluir registros por falta de información del país.

(JOIN 2 - INNER JOIN con Agrupación): Usando el mismo JOIN del ejercicio anterior, averigua: ¿Cuáles son los 5 países (muestra el nombre_pais) que tienen la mayor cantidad de reservas canceladas?


```python
%sql
SELECT
    p.nombre_pais,
    COUNT(*) AS total_cancelaciones
FROM cursodataacademy.proyecto_sql_vnrp.tb_reservas AS r
INNER JOIN cursodataacademy.proyecto_sql_vnrp.tb_paises AS p
    ON r.country = p.cod_pais
WHERE r.is_canceled = 1
GROUP BY p.nombre_pais
ORDER BY total_cancelaciones DESC, p.nombre_pais ASC
LIMIT 5;
```

**Conclusiones**
Portugal lidera con 27 519 cancelaciones, seguido de Reino Unido (2 453), España (2 177), Francia (1 934) e Italia (1 333). Portugal merece atención prioritaria por su volumen.

**(JOIN 3 - Simulación FULL OUTER JOIN): Supongamos que Recursos Humanos tiene una tabla tb_vendedores (con id_vendedor y nombre_vendedor) y cruzamos las reservas para saber qué vendedor atendió qué reserva. Realiza un FULL OUTER JOIN entre las reservas y los vendedores. Selecciona el canal de distribución y el nombre del vendedor para identificar si hay reservas sin vendedor o vendedores sin reservas asignadas.**


```python
%sql
CREATE OR REPLACE TABLE cursodataacademy.proyecto_sql_vnrp.tb_vendedores AS
SELECT *
FROM VALUES
    (1, 'Ana Torres'),
    (2, 'Luis Pérez'),
    (3, 'María López')
AS vendedores(id_vendedor, nombre_vendedor);
```

```python
%sql
CREATE OR REPLACE TABLE cursodataacademy.proyecto_sql_vnrp.tb_reservas_simuladas AS
SELECT *
FROM VALUES
    (1, 'TA/TO', 1),
    (2, 'Direct', 2),
    (3, 'Corporate', CAST(NULL AS INT))
AS reservas(id_reserva, distribution_channel, id_vendedor);
```

```python
%sql
SELECT
    R.distribution_channel,
    V.nombre_vendedor
FROM cursodataacademy.proyecto_sql_vnrp.tb_reservas_simuladas AS R
FULL OUTER JOIN cursodataacademy.proyecto_sql_vnrp.tb_vendedores AS V
    ON R.id_vendedor = V.id_vendedor;
```

**Conclusiones**
En los datos simulados se identificó una reserva del canal Corporate sin vendedor asignado y una vendedora, María López, sin reservas asociadas. El cruce permite detectar ambos casos para revisar la asignación comercial.

**(JOIN 4 - RIGHT JOIN): El departamento de contabilidad tiene una tabla de facturación (tb_facturas con id_reserva y monto_facturado). Usando un RIGHT JOIN desde las reservas hacia las facturas, garantiza que veamos todas las facturas emitidas, cruzadas con el tipo de hotel (hotel).**

```python
%sql
CREATE OR REPLACE TABLE cursodataacademy.proyecto_sql_vnrp.tb_reservas_simuladas AS
SELECT *
FROM VALUES
    (1, 'City Hotel', 'TA/TO', 1),
    (2, 'Resort Hotel', 'Direct', 2),
    (3, 'City Hotel', 'Corporate', CAST(NULL AS INT))
AS reservas(id_reserva, hotel, distribution_channel, id_vendedor);

CREATE OR REPLACE TABLE cursodataacademy.proyecto_sql_vnrp.tb_facturas AS
SELECT *
FROM VALUES
    (1, 250.00),
    (2, 480.00),
    (4, 150.00)
AS facturas(id_reserva, monto_facturado);
```

```python
%sql
SELECT
    f.id_reserva,
    f.monto_facturado,
    r.hotel
FROM cursodataacademy.proyecto_sql_vnrp.tb_reservas_simuladas AS r
RIGHT JOIN cursodataacademy.proyecto_sql_vnrp.tb_facturas AS f
    ON r.id_reserva = f.id_reserva;
```

**Conclusiones**
En la simulación se conservaron las tres facturas, incluida la de USD 150 correspondiente al identificador de reserva 4, sin coincidencia en reservas. Este registro requiere conciliación para identificar el hotel asociado y verificar su trazabilidad.
